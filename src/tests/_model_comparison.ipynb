```python
plt.figure(figsize = (9,6))
plt.plot(nglogll_mat_nbrgds['Training'], label = 'NBRGDS')
plt.plot(nglogll_mat_gs_nbrgds['Training'], label = 'GS-NBRGDS')
plt.plot(nglogll_mat_fs_nbrgds['Training'], label = 'FS-NBRGDS')
plt.plot(nglogll_mat_prgds['Training'], label = 'PRGDS')
plt.plot(nglogll_mat_gs_prgds['Training'], label = 'GS-PRGDS' )
plt.plot(nglogll_mat_fs_prgds['Training'], label = 'FS-PRGDS' )
plt.plot(nglogll_mat_pgds['Training'], label = 'PGDS' )
plt.plot(nglogll_mat_gs_pgds['Training'], label = 'GS-PGDS' )
plt.plot(nglogll_mat_fs_pgds['Training'], label = 'FS-PGDS' )

plt.xscale('log')
plt.legend()
plt.title('Negative log likelihood over training period')
plt.show()
```

```python
import pickle 
import numpy as np
import matplotlib.pyplot as plt

with open('icew_nglogll_tensor_gs_nbrgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_gs_nbrgds = pickle.load(f)

with open('icew_nglogll_tensor_fs_nbrgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_fs_nbrgds = pickle.load(f)

with open('icew_nglogll_tensor_prgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_prgds = pickle.load(f)  

with open('icew_nglogll_tensor_pgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_pgds = pickle.load(f) 

with open('icew_nglogll_tensor_gs_prgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_gs_prgds = pickle.load(f)  

with open('icew_nglogll_tensor_fs_prgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_fs_prgds = pickle.load(f)  

with open('icew_nglogll_tensor_nbrgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_nbrgds = pickle.load(f)  

with open('icew_nglogll_tensor_fs_pgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_fs_pgds = pickle.load(f)  

with open('icew_nglogll_tensor_gs_pgds_eps01_800_200.pkl', 'rb') as f:
    nglogll_tensor_gs_pgds = pickle.load(f)  
```

```python
color1 = (11/255, 85/255, 159/255)
color2 = (128/255, 128/255, 128/255)
color3 = (180/255, 4/255, 38/255)

fig, axs = plt.subplots(2,2, figsize = (9,8))
axs[0,0].plot(np.arange(1,1001),nglogll_tensor_nbrgds['Training'], label = 'NBRGDS', color = color1, alpha = 1)
axs[0,0].plot(np.arange(1,1001),nglogll_tensor_prgds['Training'], label = 'PRGDS', color = color2, alpha = 1)
axs[0,0].plot(np.arange(1,1001),nglogll_tensor_pgds['Training'], label = 'PGDS', color = color3, alpha = 1)
axs[0,0].set_xscale('log')
axs[0,0].legend()

axs[0,1].plot(np.arange(1,1001),nglogll_tensor_nbrgds['Training'], label = 'NBRGDS', color = color1, alpha = 1)
axs[0,1].plot(np.arange(1,1001),nglogll_tensor_gs_nbrgds['Training'], label = 'DGS-NBRGDS', color = color2, alpha = 1)
axs[0,1].plot(np.arange(1,1001),nglogll_tensor_fs_nbrgds['Training'], label = 'DFS-NBRGDS', color = color3, alpha = 1)
axs[0,1].set_xscale('log')
axs[0,1].legend()

axs[1,0].plot(np.arange(1,1001),nglogll_tensor_prgds['Training'], label = 'PRGDS', color = color1, alpha = 1)
axs[1,0].plot(np.arange(1,1001),nglogll_tensor_gs_prgds['Training'], label = 'DGS-PRGDS', color = color2, alpha = 1)
axs[1,0].plot(np.arange(1,1001),nglogll_tensor_fs_prgds['Training'], label = 'DFS-PRGDS', color = color3, alpha = 1)
axs[1,0].set_xscale('log')
axs[1,0].legend()

axs[1,1].plot(np.arange(1,1001),nglogll_tensor_pgds['Training'], label = 'PGDS', color = color1, alpha = 1)
axs[1,1].plot(np.arange(1,1001),nglogll_tensor_gs_pgds['Training'], label = 'DGS-PGDS', color = color2, alpha = 1)
axs[1,1].plot(np.arange(1,1001),nglogll_tensor_fs_pgds['Training'], label = 'DFS-PGDS', color = color3, alpha = 1)
axs[1,1].set_xscale('log')
axs[1,1].legend()

for ax in axs.flat:
    ax.legend(prop={'size': 15})  # 设置图例的字号为12
    ax.tick_params(axis='both', which='major', labelsize=15)  # 设置主刻度的字号为12
    ax.yaxis.set_major_formatter(plt.FormatStrFormatter('%.2f'))

    # 设置纵轴上只显示5个刻度
    y_min, y_max = ax.get_ylim()
    ax.set_yticks(np.linspace(y_min + 0.01, y_max-0.01, 5))

plt.show()
```

# iteration: 200+200

```python
%matplotlib inline
fig, axs = plt.subplots(50, 2, figsize = (16,200))
for i, ax in enumerate(axs.flatten()):
    ax.plot(model.the.T[i])
    ax.set_title('Component {}'.format(i+1))
plt.show()
```

```python
%matplotlib inline
transition_steady = torch.stack(transition_collection) 
transition_mean = torch.mean(transition_steady, axis = 0)
fig, axs = plt.subplots(1, 5, figsize=(18,6))
for i, ax in enumerate(axs.flatten()):
    im = ax.imshow(transition_mean[i], cmap='jet')
    ax.set_title(f'Transition Matrix: Stage {i+1}')
    # Add colorbar to each subplot
    cbar = plt.colorbar(im, ax=ax, shrink=0.22)
plt.tight_layout()
plt.show()

%matplotlib inline
fig, axs = plt.subplots(1,5, figsize=(18,12))
for i,ax in enumerate(axs.flatten()):
    Z = model.Z[i]
    im = ax.imshow(Z+Z.T, cmap='jet')
    cbar = plt.colorbar(im, ax=ax, shrink=0.22)
    ax.set_title(f'Cumulative Adjacent Matrix: Stage{i+1}')
plt.tight_layout()
plt.show()

%matplotlib inline
fig, axs = plt.subplots(1,5, figsize = (18,12))
for i, ax in enumerate(axs.flatten()):
    ProbAve = model.ProbAve[i] 
    im = ax.imshow(ProbAve + ProbAve.T, cmap='jet')
    ax.set_title(f'Link ProbAve:Stage{i+1}')
    cbar = plt.colorbar(im, ax = ax, shrink=0.205)
plt.tight_layout()
plt.show()
```
